```python
import mysql.connector

# Connect to MySQL
mydb = mysql.connector.connect(
    host="localhost",         # or "127.0.0.1"
    user="root",     # e.g., "root"
    password="1234",
    database="sakila"
)

# Create cursor
cursor = mydb.cursor()

# Run a sample query
cursor.execute("SHOW TABLES;")

# Fetch and print results
for table in cursor:
    print(table)

# Close connection when done
cursor.close()
mydb.close()
```

## mysql.connector.connect() 

* creates a connection object to the MySQL server.

* host: The address of the MySQL server (localhost = your own computer).

* user and password: Your MySQL login credentials.

* database: The database you want to use (in this case, sakila).

## cursor = mydb.cursor()

* A cursor is an object you create from the connection.

* cursor is a database control structure used to:

* Send SQL queries to the MySQL server.

* Fetch results (like rows or column metadata).

## cursor.close()

## mydb.close()

* Frees up system/database resources.

* Always close both cursor and connection when done.

```python
import pandas as pd

def run_query(query):
    cursor.execute(query)
    columns = [col[0] for col in cursor.description]
    rows = cursor.fetchall()
    return pd.DataFrame(rows, columns=columns)

# # Example usage:
# query1 = "SELECT * FROM employees;"
# df_employees = run_query(query1)
# print(df_employees)
```

```python
# 1. List all employees
df1 = run_query("SELECT * FROM employees;")

# 2. List all departments
df2 = run_query("SELECT * FROM departments;")

# 3. Employees with their department names
df3 = run_query("""
SELECT e.emp_name AS employee_name, d.dept_name AS department_name
FROM employees e
JOIN departments d ON e.dept_id = d.dept_id;
""")

# 4. Count of employees in each department
df4 = run_query("""
SELECT d.dept_name AS department, COUNT(e.emp_id) AS employee_count
FROM employees e
JOIN departments d ON e.dept_id = d.dept_id
GROUP BY d.dept_name;
""")

# 5. Employees in a specific department (e.g., HR)
df5 = run_query("""
SELECT e.emp_name
FROM employees e
JOIN departments d ON e.dept_id = d.dept_id
WHERE d.dept_name = 'HR';
""")

# 6. Departments with no employees
df6 = run_query("""
SELECT d.dept_name
FROM departments d
LEFT JOIN employees e ON e.dept_id = d.dept_id
WHERE e.dept_id IS NULL;
""")

# 7. Top 5 highest-paid employees (if salary exists)
df7 = run_query("""
SELECT emp_name, emp_salary
FROM employees
ORDER BY emp_salary DESC
LIMIT 5;
""")


```
